````python
import pandas as pd

manifest = pd.read_csv("data/interim/manifest.csv")
cycles = pd.read_csv("data/interim/cycles.csv")
print(manifest.head())
print(cycles["class_label"].value_counts())
print(cycles.groupby(["official_partition", "patient_id"]).size())
````

## Patient-independent split integrity

No patient may appear in more than one partition.

````python
grouped = manifest.groupby("patient_id")["official_partition"].nunique()
assert (grouped == 1).all(), "Patient leakage detected!"
print("Split integrity OK")
````
