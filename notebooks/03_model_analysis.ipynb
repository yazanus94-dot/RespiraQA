```python
import json
import pandas as pd

with open("models/model_metadata.json", encoding="utf-8") as fh:
    metadata = json.load(fh)
print("Feature count:", len(metadata["feature_names"]))
print("Classes:", metadata["classes"])
print("Training details:", metadata["training"])

predictions = pd.read_csv("reports/metrics/predictions.csv")
predictions.head()
```

```python
from respiraqa.models.baseline import load_model
import pandas as pd

model, metadata = load_model("models/respiraqa_baseline.joblib",
                             "models/model_metadata.json")
features = pd.read_csv("data/processed/features.csv")
names = metadata["feature_names"]
importances = model.named_steps["classifier"].feature_importances_
top = pd.Series(importances, index=names).sort_values(ascending=False).head(10)
top.plot(kind="barh", figsize=(8, 5), title="Top 10 feature importances");
```
