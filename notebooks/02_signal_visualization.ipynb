# 02 — Signal Visualization

Visualize raw waveforms, filtered signals, and spectrograms for an
annotated respiratory cycle.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from respiraqa.data.audio_loader import load_audio
from respiraqa.preprocessing.filtering import bandpass_filter
from respiraqa.preprocessing.segmentation import extract_cycle

cycles = pd.read_csv("data/interim/cycles.csv")
row = cycles.iloc[0]
audio, info = load_audio(row["audio_path"], target_sample_rate=None, mono=True)
cycle = extract_cycle(audio, int(row["sample_rate"]), row["start_time"], row["end_time"])
filtered = bandpass_filter(cycle, int(row["sample_rate"]), 100.0,
                           min(2000.0, row["sample_rate"] / 2 * 0.95))

fig, axes = plt.subplots(3, 1, figsize=(10, 8))
t = np.arange(cycle.size) / row["sample_rate"]
axes[0].plot(t, cycle); axes[0].set_title(f"Raw cycle ({row['class_label']})")
axes[1].plot(t, filtered); axes[1].set_title("Band-pass filtered")
axes[2].specgram(cycle, NFFT=256, Fs=row["sample_rate"]); axes[2].set_title("Spectrogram")
plt.tight_layout(); plt.show()
```
